```python
# Ejercicio 11: Ordenamiento Básico
# Solicita: código de vendedor, número de factura, importe de ventas.
# Finaliza con código -1.
# Ordena por código de vendedor (mayor a menor) usando una función específica.

def ordenamiento_burbuja(codigos, facturas, importes):
    """Ordena las listas paralelas por código de vendedor (mayor a menor)."""
    n = len(codigos)
    for i in range(n):
        for j in range(0, n - i - 1):
            if codigos[j] < codigos[j + 1]:
                # Intercambiar en todas las listas
                codigos[j], codigos[j + 1] = codigos[j + 1], codigos[j]
                facturas[j], facturas[j + 1] = facturas[j + 1], facturas[j]
                importes[j], importes[j + 1] = importes[j + 1], importes[j]

# Listas paralelas
codigos_vendedor = []
numeros_factura = []
importes_venta = []

print("=== INGRESO DE VENTAS ===")
while True:
    codigo = int(input("Ingrese código de vendedor (-1 para finalizar): "))
    if codigo == -1:
        break
    
    factura = int(input("Ingrese número de factura: "))
    importe = float(input("Ingrese importe de venta: "))
    
    codigos_vendedor.append(codigo)
    numeros_factura.append(factura)
    importes_venta.append(importe)

# Ordenar usando la función
ordenamiento_burbuja(codigos_vendedor, numeros_factura, importes_venta)

# Mostrar resultados
print("\n=== LISTADO ORDENADO POR CÓDIGO (MAYOR A MENOR) ===")
print(f"{'Código':<10} {'Factura':<10} {'Importe':<10}")
print("-" * 30)
for i in range(len(codigos_vendedor)):
    print(f"{codigos_vendedor[i]:<10} {numeros_factura[i]:<10} ${importes_venta[i]:<9.2f}")
```

```python
# Ejercicio 12: Ordenamiento por Selección e Inserción
# Mismo enunciado del Ejercicio 11, pero implementando selección e inserción.

def ordenamiento_seleccion(codigos, facturas, importes):
    """Ordena por selección de mayor a menor."""
    n = len(codigos)
    for i in range(n):
        max_idx = i
        for j in range(i + 1, n):
            if codigos[j] > codigos[max_idx]:
                max_idx = j
        
        # Intercambiar
        codigos[i], codigos[max_idx] = codigos[max_idx], codigos[i]
        facturas[i], facturas[max_idx] = facturas[max_idx], facturas[i]
        importes[i], importes[max_idx] = importes[max_idx], importes[i]

def ordenamiento_insercion(codigos, facturas, importes):
    """Ordena por inserción de mayor a menor."""
    n = len(codigos)
    for i in range(1, n):
        codigo_clave = codigos[i]
        factura_clave = facturas[i]
        importe_clave = importes[i]
        
        j = i - 1
        while j >= 0 and codigos[j] < codigo_clave:
            codigos[j + 1] = codigos[j]
            facturas[j + 1] = facturas[j]
            importes[j + 1] = importes[j]
            j -= 1
        
        codigos[j + 1] = codigo_clave
        facturas[j + 1] = factura_clave
        importes[j + 1] = importe_clave

# Listas paralelas
codigos_vendedor = []
numeros_factura = []
importes_venta = []

print("=== INGRESO DE VENTAS ===")
while True:
    codigo = int(input("Ingrese código de vendedor (-1 para finalizar): "))
    if codigo == -1:
        break
    
    factura = int(input("Ingrese número de factura: "))
    importe = float(input("Ingrese importe de venta: "))
    
    codigos_vendedor.append(codigo)
    numeros_factura.append(factura)
    importes_venta.append(importe)

print("\n¿Qué método desea usar?")
print("1) Selección")
print("2) Inserción")
metodo = int(input("Seleccione (1 o 2): "))

if metodo == 1:
    ordenamiento_seleccion(codigos_vendedor, numeros_factura, importes_venta)
    print("\n=== LISTADO ORDENADO POR SELECCIÓN ===")
else:
    ordenamiento_insercion(codigos_vendedor, numeros_factura, importes_venta)
    print("\n=== LISTADO ORDENADO POR INSERCIÓN ===")

print(f"{'Código':<10} {'Factura':<10} {'Importe':<10}")
print("-" * 30)
for i in range(len(codigos_vendedor)):
    print(f"{codigos_vendedor[i]:<10} {numeros_factura[i]:<10} ${importes_venta[i]:<9.2f}")
```

```python
# Ejercicio 13: Búsqueda Binaria
# Leer datos de ventas hasta ingresar -1, ordenarlos, y permitir búsqueda binaria por código.

def ordenamiento_burbuja_desc(codigos, facturas, importes):
    """Ordena de mayor a menor."""
    n = len(codigos)
    for i in range(n):
        for j in range(0, n - i - 1):
            if codigos[j] < codigos[j + 1]:
                codigos[j], codigos[j + 1] = codigos[j + 1], codigos[j]
                facturas[j], facturas[j + 1] = facturas[j + 1], facturas[j]
                importes[j], importes[j + 1] = importes[j + 1], importes[j]

def busqueda_binaria(codigos, facturas, importes, codigo_buscado):
    """Realiza búsqueda binaria y retorna (encontrado, índice, factura, importe)."""
    izq = 0
    der = len(codigos) - 1
    
    while izq <= der:
        mid = (izq + der) // 2
        
        if codigos[mid] == codigo_buscado:
            return True, mid, facturas[mid], importes[mid]
        elif codigos[mid] > codigo_buscado:
            izq = mid + 1
        else:
            der = mid - 1
    
    return False, -1, None, None

# Listas paralelas
codigos_vendedor = []
numeros_factura = []
importes_venta = []

print("=== INGRESO DE VENTAS ===")
while True:
    codigo = int(input("Ingrese código de vendedor (-1 para finalizar): "))
    if codigo == -1:
        break
    
    factura = int(input("Ingrese número de factura: "))
    importe = float(input("Ingrese importe de venta: "))
    
    codigos_vendedor.append(codigo)
    numeros_factura.append(factura)
    importes_venta.append(importe)

# Ordenar datos
ordenamiento_burbuja_desc(codigos_vendedor, numeros_factura, importes_venta)

print("\n=== DATOS ORDENADOS ===")
print(f"{'Código':<10} {'Factura':<10} {'Importe':<10}")
print("-" * 30)
for i in range(len(codigos_vendedor)):
    print(f"{codigos_vendedor[i]:<10} {numeros_factura[i]:<10} ${importes_venta[i]:<9.2f}")

# Búsqueda binaria
print("\n=== BÚSQUEDA BINARIA ===")
while True:
    codigo_busca = int(input("\nIngrese código de vendedor a buscar (-1 para salir): "))
    if codigo_busca == -1:
        break
    
    encontrado, idx, factura, importe = busqueda_binaria(codigos_vendedor, numeros_factura, importes_venta, codigo_busca)
    
    if encontrado:
        print(f"✓ Vendedor encontrado:")
        print(f"  Código: {codigo_busca}")
        print(f"  Factura: {factura}")
        print(f"  Importe: ${importe:.2f}")
    else:
        print(f"✗ El vendedor con código {codigo_busca} no se encuentra registrado.")

print("\nFin de la búsqueda.")
```
