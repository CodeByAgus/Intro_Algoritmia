```python
# Ejercicio 1 - SOLUCIÓN CON DOS LISTAS
legajos = []
notas = []

# Carga de datos
print("=== CARGA DE DATOS ===")
while True:
    legajo = int(input("Ingrese legajo (-1 para finalizar): "))
    if legajo == -1:
        break
    
    while True:
        nota = int(input(f"Ingrese nota para legajo {legajo} (1-10): "))
        if 1 <= nota <= 10:
            break
        print("Nota inválida. Debe estar entre 1 y 10.")
    
    legajos.append(legajo)
    notas.append(nota)

# Cálculos
aprobados = sum(1 for n in notas if n >= 4)
desaprobados = sum(1 for n in notas if n < 4)
promedio = sum(notas) / len(notas) if notas else 0
legajos_superior = [legajos[i] for i in range(len(legajos)) if notas[i] > promedio]

print("\n=== RESULTADOS ===")
print(f"Alumnos aprobados: {aprobados}")
print(f"Alumnos desaprobados: {desaprobados}")
print(f"Promedio de notas: {promedio:.2f}")
print(f"Legajos que superan el promedio: {legajos_superior}")

# Ordenar por legajo
indices_ordenados = sorted(range(len(legajos)), key=lambda i: legajos[i])
print("\n=== LISTADO ORDENADO POR LEGAJO ===")
for i in indices_ordenados:
    print(f"Legajo: {legajos[i]}, Nota: {notas[i]}")

# Búsqueda binaria
print("\n=== BÚSQUEDA BINARIA ===")
legajos_ordenados = sorted(legajos)
notas_por_legajo = {legajos[i]: notas[i] for i in range(len(legajos))}

while True:
    legajo_busca = int(input("Ingrese legajo a buscar (-1 para terminar): "))
    if legajo_busca == -1:
        break
    
    # Búsqueda binaria
    izq, der = 0, len(legajos_ordenados) - 1
    encontrado = False
    
    while izq <= der:
        mid = (izq + der) // 2
        if legajos_ordenados[mid] == legajo_busca:
            print(f"Legajo {legajo_busca}: Nota = {notas_por_legajo[legajo_busca]}")
            encontrado = True
            break
        elif legajos_ordenados[mid] < legajo_busca:
            izq = mid + 1
        else:
            der = mid - 1
    
    if not encontrado:
        print(f"Legajo {legajo_busca} no encontrado.")

# SOLUCIÓN CON MATRIZ DE DOS FILAS
print("\n\n=== SOLUCIÓN CON MATRIZ ===")
matriz = [[], []]

print("=== CARGA DE DATOS ===")
while True:
    legajo = int(input("Ingrese legajo (-1 para finalizar): "))
    if legajo == -1:
        break
    
    while True:
        nota = int(input(f"Ingrese nota para legajo {legajo} (1-10): "))
        if 1 <= nota <= 10:
            break
        print("Nota inválida. Debe estar entre 1 y 10.")
    
    matriz[0].append(legajo)
    matriz[1].append(nota)

# Cálculos con matriz
aprobados = sum(1 for n in matriz[1] if n >= 4)
desaprobados = sum(1 for n in matriz[1] if n < 4)
promedio = sum(matriz[1]) / len(matriz[1]) if matriz[1] else 0
legajos_superior = [matriz[0][i] for i in range(len(matriz[0])) if matriz[1][i] > promedio]

print("\n=== RESULTADOS ===")
print(f"Alumnos aprobados: {aprobados}")
print(f"Alumnos desaprobados: {desaprobados}")
print(f"Promedio de notas: {promedio:.2f}")
print(f"Legajos que superan el promedio: {legajos_superior}")

# Ordenar matriz por legajo
indices_ordenados = sorted(range(len(matriz[0])), key=lambda i: matriz[0][i])
print("\n=== LISTADO ORDENADO POR LEGAJO ===")
for i in indices_ordenados:
    print(f"Legajo: {matriz[0][i]}, Nota: {matriz[1][i]}")

# Búsqueda binaria con matriz
print("\n=== BÚSQUEDA BINARIA ===")
legajos_ordenados = sorted(matriz[0])
notas_por_legajo = {matriz[0][i]: matriz[1][i] for i in range(len(matriz[0]))}

while True:
    legajo_busca = int(input("Ingrese legajo a buscar (-1 para terminar): "))
    if legajo_busca == -1:
        break
    
    # Búsqueda binaria
    izq, der = 0, len(legajos_ordenados) - 1
    encontrado = False
    
    while izq <= der:
        mid = (izq + der) // 2
        if legajos_ordenados[mid] == legajo_busca:
            print(f"Legajo {legajo_busca}: Nota = {notas_por_legajo[legajo_busca]}")
            encontrado = True
            break
        elif legajos_ordenados[mid] < legajo_busca:
            izq = mid + 1
        else:
            der = mid - 1
    
    if not encontrado:
        print(f"Legajo {legajo_busca} no encontrado.")
```

```python
# Ejercicio 2 - Sistema de Consorcios
unidades = []
superficies = []
VALOR_POR_M2 = 50  # valor fijo de expensas por m² por mes

print("=== CARGA DE DATOS ===")
for i in range(20):
    while True:
        num_unidad = int(input(f"Ingrese número de unidad {i+1} (o -1 para finalizar): "))
        if num_unidad == -1:
            break
        if num_unidad not in unidades:  # Validar no duplicados
            unidades.append(num_unidad)
            superficie = float(input(f"Ingrese superficie de unidad {num_unidad} (m²): "))
            superficies.append(superficie)
            break
        else:
            print("Error: Esta unidad ya fue ingresada.")
    
    if num_unidad == -1:
        break

# a) Calcular promedio de expensas
expensas = [s * VALOR_POR_M2 for s in superficies]
promedio_expensas = sum(expensas) / len(expensas) if expensas else 0

print(f"\na) Promedio de expensas del mes: ${promedio_expensas:.2f}")

# b) Ordenar de mayor a menor según superficie
indices_ordenados = sorted(range(len(unidades)), key=lambda i: superficies[i], reverse=True)

print("\nb) Listado ordenado por superficie (mayor a menor):")
print("Número de Unidad | Superficie (m²) | Expensa ($)")
print("-" * 50)
for i in indices_ordenados:
    print(f"{unidades[i]:^16} | {superficies[i]:^15.2f} | {expensas[i]:^11.2f}")
```

```python
# Ejercicio 3 - Adivina el Número
import random

historial = []  # Lista para guardar mejores puntajes

def juego():
    numero_secreto = random.randint(1000, 9999)
    intentos = 0
    
    print("\n=== ADIVINA EL NÚMERO (4 cifras) ===")
    print("Ingresa tu intento o -1 para abandonar.")
    
    while True:
        intento = int(input("Tu intento: "))
        
        if intento == -1:
            print(f"Abandonaste el juego después de {intentos} intentos.")
            return None
        
        intentos += 1
        
        if intento == numero_secreto:
            print(f"\n¡¡¡ GANASTE !!! Adivinaste en {intentos} intentos.")
            return intentos
        elif intento < numero_secreto:
            print("El número secreto es MAYOR.")
        else:
            print("El número secreto es MENOR.")

# Loop principal
while True:
    resultado = juego()
    
    if resultado is not None:
        # Verificar si es un nuevo récord
        if not historial or resultado < historial[-1][1]:
            nombre = input("¡Nuevo récord! Ingrese su DNI: ")
            historial.append((nombre, resultado))
            historial.sort(key=lambda x: x[1])
            historial = historial[:5]  # Mantener solo los 5 mejores
    
    # Mostrar los 5 mejores puntajes
    if historial:
        print("\n=== TOP 5 MEJORES PUNTAJES ===")
        for idx, (dni, puntaje) in enumerate(historial, 1):
            print(f"{idx}. DNI: {dni} - Intentos: {puntaje}")
    
    # Preguntar si juega otra vez
    jugar_nuevamente = input("\n¿Desea jugar otra vez? (s/n): ").lower()
    if jugar_nuevamente != 's':
        print("¡Gracias por jugar!")
        break
```

```python
# Ejercicio 4 - Adivina el Número CON PISTAS MEJORADAS
import random

def contar_digitos(numero_secreto, intento):
    """Retorna (correctos, aproximados)"""
    secreto_str = str(numero_secreto).zfill(4)
    intento_str = str(intento).zfill(4)
    
    correctos = sum(1 for i in range(4) if secreto_str[i] == intento_str[i])
    
    # Para aproximados, contar dígitos que están en el número pero en posición incorrecta
    aproximados = 0
    for i in range(4):
        if secreto_str[i] != intento_str[i] and intento_str[i] in secreto_str:
            aproximados += 1
    
    return correctos, aproximados

def juego_mejorado():
    numero_secreto = random.randint(1000, 9999)
    intentos = 0
    
    print("\n=== ADIVINA EL NÚMERO (4 cifras) ===")
    print("Ingresa tu intento o -1 para abandonar.")
    print("Recibirás pistas: M dígitos correctos y N dígitos aproximados.\n")
    
    while True:
        intento = int(input("Tu intento: "))
        
        if intento == -1:
            print(f"Abandonaste el juego después de {intentos} intentos.")
            print(f"El número secreto era: {numero_secreto}")
            return None
        
        intentos += 1
        
        if intento == numero_secreto:
            print(f"\n¡¡¡ GANASTE !!! Adivinaste en {intentos} intentos.")
            return intentos
        else:
            correctos, aproximados = contar_digitos(numero_secreto, intento)
            print(f"Pista: {correctos} dígitos correctos y {aproximados} dígitos aproximados.")

historial = []  # Lista para guardar mejores puntajes

# Loop principal
while True:
    resultado = juego_mejorado()
    
    if resultado is not None:
        # Verificar si es un nuevo récord
        if not historial or resultado < historial[-1][1]:
            nombre = input("¡Nuevo récord! Ingrese su DNI: ")
            historial.append((nombre, resultado))
            historial.sort(key=lambda x: x[1])
            historial = historial[:5]  # Mantener solo los 5 mejores
    
    # Mostrar los 5 mejores puntajes
    if historial:
        print("\n=== TOP 5 MEJORES PUNTAJES ===")
        for idx, (dni, puntaje) in enumerate(historial, 1):
            print(f"{idx}. DNI: {dni} - Intentos: {puntaje}")
    
    # Preguntar si juega otra vez
    jugar_nuevamente = input("\n¿Desea jugar otra vez? (s/n): ").lower()
    if jugar_nuevamente != 's':
        print("¡Gracias por jugar!")
        break
```

```python
# Ejercicio 5 - Secuencias Separadas por Ceros
import random

print("=== GENERACIÓN DE SECUENCIAS ===")
N = int(input("Ingrese cantidad de números para generar: "))

# Generar números aleatorios
numeros = [random.randint(1, 20) for _ in range(N)]
print(f"Lista original: {numeros}")

# Construir lista de secuencias
secuencias = []
suma_actual = 0

for num in numeros:
    if suma_actual + num <= 20:
        secuencias.append(num)
        suma_actual += num
    else:
        secuencias.append(0)  # Separador
        secuencias.append(num)
        suma_actual = num

# Agregar 0 final
secuencias.append(0)

print(f"Resultado:     {secuencias}")
```

```python
"""Club Deportivo - Gestión de Inscripción en Torneo
Programa para cargar una lista de socios y eliminar un socio que se desiste del torneo.
"""

def cargar_socios():
    """Carga una lista de números de socios hasta que se ingresa 0."""
    lista = []
    while True:
        socio = int(input("Ingrese número de socio (0 para finalizar): "))
        if socio == 0:
            break
        lista.append(socio)
    return lista

def eliminar_socio(lista, numero_baja):
    """Elimina un socio de la lista y retorna la lista modificada."""
    if numero_baja in lista:
        posicion = lista.index(numero_baja)
        lista.remove(numero_baja)
        print(f"✔ Confirmación: El socio {numero_baja} fue eliminado. Estaba en la posición {posicion}.")
    else:
        print("Error: El socio no se encuentra registrado.")
    
    return lista

# PROGRAMA PRINCIPAL
print("--- SISTEMA DEL CLUB DEPORTIVO ---")
listado_socios = cargar_socios()

print("\nLista original de socios:", listado_socios)

if len(listado_socios) > 0:
    socio_a_borrar = int(input("\nIngrese el número de socio que desea dar de baja: "))
    listado_socios = eliminar_socio(listado_socios, socio_a_borrar)
    print("\nLista final de inscritos:", listado_socios)
else:
    print("No se ingresaron socios al torneo.")
```
