```python
# Ejercicio 1:
# Crear una matriz de 3x4 (3 filas y 4 columnas) con valores creados al azar entre 1 y 10.
# Mostrar la matriz creada respetando el formato de 3 filas y 4 columnas por pantalla.
import random

rows, cols = 3, 4
matrix = []

for i in range(rows):
    fila = []
    for j in range(cols):
        fila.append(random.randint(1, 10))
    matrix.append(fila)

for fila in matrix:
    linea = ""
    for val in fila:
        linea += f"{val:2d} "
    print(linea)
```
