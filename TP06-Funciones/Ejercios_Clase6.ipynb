```python
# Ejercicio 1: Diseñar una función que reciba dos 
# parámetros numéricos enteros, calcule y devuelva 
# el resultado de la multiplicación de ambos utilizando 
# sólo sumas.

def multipicacion(a,b):
    total = 0
    for i in range (b):
        total += a
    return total

a = int(input("ingrese numero"))
b= int(input("ingrese multiplicador"))
multi = multipicacion(a,b)
print(multi)
```
