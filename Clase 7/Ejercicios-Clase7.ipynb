```python
def ingresa_entero(minimo,maximo):
    numero= int(input("Ingrese entero:"))
    if numero > maximo :
        r= ingresa_entero(minimo, maximo)
    elif numero < minimo:
        r= ingresa_entero(minimo, maximo)
    else:
        r= numero
    return r

ingresa_entero(4,6)
```
