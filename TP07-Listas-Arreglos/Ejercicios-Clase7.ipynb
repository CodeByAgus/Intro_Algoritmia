````python
# Ejercicio 8: búsqueda secuencial en lista desordenada

def busqueda_secuencial(lista, valor):
    for i, elemento in enumerate(lista):
        if elemento == valor:
            return i
    return -1

# Ejemplo de uso
lista_ej8 = [7, 2, 9, 4, 3]
print("Ejercicio 8:", busqueda_secuencial(lista_ej8, 9))
print("Ejercicio 8 (no encontrado):", busqueda_secuencial(lista_ej8, 5))
````

````python
# Ejercicio 9: ordenar tres listas en forma ascendente con selección, inserción y burbujeo

def orden_seleccion(lista):
    lista = lista.copy()
    n = len(lista)
    for i in range(n - 1):
        minimo = i
        for j in range(i + 1, n):
            if lista[j] < lista[minimo]:
                minimo = j
        lista[i], lista[minimo] = lista[minimo], lista[i]
    return lista


def orden_insercion(lista):
    lista = lista.copy()
    for i in range(1, len(lista)):
        clave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > clave:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = clave
    return lista


def orden_burbujeo(lista):
    lista = lista.copy()
    n = len(lista)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

lista_a = [9, 5, 2, 8, 1]
lista_b = [7, 4, 3, 6, 0]
lista_c = [10, 2, 8, 1, 5]

print("Ejercicio 9:")
print("Selección:", orden_seleccion(lista_a))
print("Inserción:", orden_insercion(lista_b))
print("Burbujeo:", orden_burbujeo(lista_c))
print("Para orden descendente, invierta el criterio de comparación o use lista[::-1] después de ordenar ascendente.")
````

````python
# Ejercicio 10: crear lista al azar, ordenar por inserción y buscar con búsqueda binaria
import random

def orden_insercion(lista):
    """
    Ordenamiento por inserción:
    - Recorre la lista desde el segundo elemento (i=1)
    - Para cada elemento, lo compara con los anteriores
    - Si encuentra un elemento mayor, lo desplaza a la derecha
    - Inserta el elemento en su posición correcta
    - Complejidad: O(n²) en peor caso
    """
    lista = lista.copy()
    for i in range(1, len(lista)):
        clave = lista[i]  # Elemento a insertar
        j = i - 1  # Posición del último elemento ordenado

        # Desplaza elementos mayores a la clave hacia la derecha
        while j >= 0 and lista[j] > clave:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = clave  # Inserta la clave en su posición correcta
    return lista

def busqueda_binaria(lista, valor):
    """
    Búsqueda binaria en lista ordenada:
    - Divide el espacio de búsqueda a la mitad en cada iteración
    - Si el valor está en el medio, lo encontró
    - Si es menor, busca en la mitad izquierda
    - Si es mayor, busca en la mitad derecha
    - Retorna el índice o -1 si no lo encuentra
    """
    izquierda, derecha = 0, len(lista) - 1
    while izquierda <= derecha:
        medio = (izquierda + derecha) // 2
        if lista[medio] == valor:
            return medio
        elif lista[medio] < valor:
            izquierda = medio + 1
        else:
            derecha = medio - 1
    return -1

# Crear lista con números al azar
lista_ej10 = [random.randint(0, 100) for _ in range(10)]
print("Ejercicio 10: lista original:", lista_ej10)

# Ordenar usando inserción
lista_ej10 = orden_insercion(lista_ej10)
print("Lista ordenada por inserción:", lista_ej10)

# Búsqueda interactiva
exitos = 0
no_encontrados = 0
while True:
    buscado = int(input("\nIngrese valor para buscar (-1 para terminar): "))
    if buscado == -1:
        break
    if busqueda_binaria(lista_ej10, buscado) != -1:
        exitos += 1
        print("✓ Encontrado.")
    else:
        no_encontrados += 1
        print("✗ No se encontró.")

print("\n--- Resultados ---")
print("Búsquedas exitosas:", exitos)
print("Búsquedas no encontradas:", no_encontrados)
````

## Explicación del Método de Inserción

**Idea principal:** Construye la lista ordenada elemento por elemento, como si ordenaras cartas en tu mano.

### Pasos del algoritmo:

1. **Comienza desde el segundo elemento** (índice 1) porque el primer elemento ya está "ordenado"
2. **Para cada elemento:**
   - Lo llama "clave" - es el que vamos a insertar en la parte ya ordenada
   - Compara la clave con los elementos anteriores (de derecha a izquierda)
   - Si encuentra un elemento **mayor** que la clave, lo desplaza a la derecha
   - Continúa hasta encontrar su posición correcta
3. **Inserta la clave** en el espacio que quedó libre

### Ejemplo visual con [5, 2, 8, 1]:

```
Inicio: [5, 2, 8, 1]

Paso 1: clave=2
- Compara 2 con 5 → 5 > 2, desplaza 5
- Inserta 2: [2, 5, 8, 1]

Paso 2: clave=8
- Compara 8 con 5 → no es mayor, inserta aquí: [2, 5, 8, 1]

Paso 3: clave=1
- Compara 1 con 8 → 8 > 1, desplaza
- Compara 1 con 5 → 5 > 1, desplaza
- Compara 1 con 2 → 2 > 1, desplaza
- Inserta 1: [1, 2, 5, 8]
```

### Ventajas:
- ✓ Eficiente con listas pequeñas
- ✓ Ordenamiento "in place" (usa poco espacio extra)
- ✓ Estable (mantiene el orden de elementos iguales)
- ✓ Complejidad: O(n²) en peor caso, O(n) en mejor caso

1. **Crea una lista** con 10 números al azar entre 0 y 100
2. **Ordena con inserción** la lista creada
3. **Permite búsquedas** con búsqueda binaria (mucho más rápida que secuencial)
4. **Finaliza** cuando ingresa -1
5. **Muestra resultados** con el total de búsquedas exitosas y no encontradas
