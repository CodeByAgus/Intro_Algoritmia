```python
# Ejercicio 1
# Leer números que representan edades de un grupo de personas,
# finalizando la lectura cuando se ingrese el número 999. Imprimir cuántos son menores de 18 años, 
# cuántos tienen 18 años o más y el promedio de edad de ambos grupos.
# Descartar valores que no representan una edad válida. (Se considera válido una edad entre 0 y 100).

# === RESOLUCIÓN CON WHILE ===
cantidad_menores = 0
cantidad_mayores = 0
suma_edades = 0

edad = int(input("Ingresa edad (999 para terminar): "))

while edad != 999:
    if edad >= 0 and edad <= 100:
        suma_edades += edad
        if edad < 18:
            cantidad_menores += 1
        else:
            cantidad_mayores += 1
       
    else:
        print("Edad no válida, descartada.")
    
    edad = int(input("Ingresa edad (999 para terminar): "))

total_personas = cantidad_menores + cantidad_mayores
if total_personas > 0:
    promedio = suma_edades / total_personas
else:
    promedio = 0

print("Menores de 18:", cantidad_menores)
print("Mayores o iguales a 18:", cantidad_mayores)
print("Promedio de edad:", promedio)
```

```python
### Ejercicio 2
# Escribir un algoritmo que permita ingresar los números de legajo de los alumnos 
# de un curso y su nota de examen final. El fin de la carga se determina ingresando un -1 en el legajo. 
# Informar para cada alumno si aprobó o no el examen considerando que se aprueba con nota mayor o 
# igual a 4. Se debe validar que la nota ingresada sea entre 1 y 10. Terminada la carga de datos, informar:
# - Cantidad de alumnos que aprobaron con nota mayor o igual a 4.
# - Cantidad de alumnos que desaprobaron el examen. Nota menor a 4.
# - Porcentaje de alumnos que están aplazados (tienen 1 en el examen).

# === RESOLUCIÓN CON WHILE ===
legajo = int(input("Ingrese Legajo del Alumno (-1 para salir): "))

aprobados = 0
desaprobados = 0
aplazos = 0
total_alumnos = 0

while legajo != -1:
    nota = int(input("Ingrese Nota del Alumno (1 a 10): "))
    if nota >= 1 and nota <= 10:
        total_alumnos += 1
        if nota >= 4:
            aprobados += 1
            print("El alumno aprobó el examen.")
        else:
            desaprobados += 1
            print("El alumno desaprobó el examen.")
            if nota == 1:
                aplazos += 1
    else:
        print("Nota inválida.")
    
    legajo = int(input("Ingrese Legajo del Alumno (-1 para salir): "))

print("Aprobados:", aprobados)
print("Desaprobados:", desaprobados)
if total_alumnos > 0:
    print("Porcentaje de aplazos:", (aplazos * 100) / total_alumnos, "%")
else:
    print("No se ingresaron alumnos.")
```

```python
# ### Ejercicio 3
# Una empresa aplica el siguiente procedimiento en la comercialización de sus productos:
# - Aplica el precio base a la primera docena de unidades.
# - Aplica un 10% de descuento a todas las unidades entre 13 y 100.
# - Aplica un 25% de descuento a todas las unidades por encima de las 100.
# Escribir un algoritmo que lea la cantidad solicitada de un producto y su precio base, y 
# muestre el valor total de la venta y el precio promedio por unidad. El fin de carga se determina
#  ingresando -1 como cantidad solicitada.
# Al finalizar informar:
# a) Cantidad de ventas realizadas total.
# b) Cantidad de ventas que se aplicaron un 10% de descuento.
# c) Cantidad de ventas que SOLO se aplicó el precio base, no se le realizaron descuentos.

# === RESOLUCIÓN CON WHILE ===
ventas_totales = 0
ventas_desc_10 = 0
ventas_sin_desc = 0

cantidad = int(input("Ingrese cantidad solicitada (-1 para salir): "))

while cantidad != -1:
    precio_base = float(input("Ingrese precio base: "))
    
    # Calcular precio por partes
    if cantidad > 100:
        precio_total = (12 * precio_base) + (88 * precio_base * 0.90) + ((cantidad - 100) * precio_base * 0.75)
    elif cantidad >= 13 and cantidad <= 100:
        precio_total = (12 * precio_base) + ((cantidad - 12) * precio_base * 0.90)
        ventas_desc_10 += 1
    else:
        precio_total = cantidad * precio_base
        ventas_sin_desc += 1
        
    precio_promedio = precio_total / cantidad if cantidad > 0 else 0
    
    print("Valor total venta:", precio_total)
    print("Precio promedio por unidad:", precio_promedio)
    
    ventas_totales += 1
    cantidad = int(input("Ingrese cantidad solicitada (-1 para salir): "))

print("Total de ventas realizadas:", ventas_totales)
print("Ventas donde maximo se aplico 10% de descuento:", ventas_desc_10)
print("Ventas sin descuentos aplicados:", ventas_sin_desc)
\n
```

```python
### Ejercicio 5
# El factorial de un número entero N mayor que cero se define como el producto de todos los enteros X tales que 0 < X <= N. Desenvolver un programa para calcular el factorial de un número dado hasta ingresar -1. Deberán rechazarse las entradas inválidas (menores que 0). Al finalizar informar cuántas veces se calculó el factorial.

# === RESOLUCIÓN CON WHILE ===
calculos = 0
n = int(input("Ingrese numero (-1 para salir): "))

while n != -1:
    if n >= 0:
        factorial = 1
        i = 1
        while i <= n:
            factorial = factorial * i
            i += 1
        print("Factorial:", factorial)
        calculos += 1
    else:
        print("Entrada inválida.")
    
    n = int(input("Ingrese numero (-1 para salir): "))

print("Veces calculado:", calculos)
\n
```
