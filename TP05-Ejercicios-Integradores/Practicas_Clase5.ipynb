```python
# Leer el conjunto de números enteros e imprimir el mayor
# El final de los datos se indica con -1
# Encontrar si hay un número máximo

# Práctica de memoria de elementos

numeromaximo = 0
numero = int(input("Ingrese número: "))
while numero != -1:
    if numero > numeromaximo:
        numeromaximo = numero
    numero = int(input("Ingrese número: "))

print(numeromaximo)


```

```python
numerofactoriar = int(input("ingrese numero a factoriar"))
factorial= (numerofactoriar - 1)
resultado = numerofactoriar
for facto in range(factorial,0,-1):
    resultado *= facto
print(resultado)
```

Pseudocodigo 

Le pido al ususario ingresar el numero a factorizar
saco el factorial que parte la multiplicacion desde un numero inferior al numero ingresado -1 
resultado siempre es el numero ingresado multiplicado por si numero inferior y este multiplicado por el segundo inferior.
se crea un bucle for para que multiplique el resultado por cada unos de los numeros posteriores al ingresado 

```python
aprobados = 0
desaprobados = 0
aplazados = 0
legajo = int(input("Ingrese legajo (-1 para terminar): "))

while legajo  !=1:  

    if legajo == -1:
        break

    nota = int(input("Ingrese nota (1-10): "))
    while nota < 1 or nota > 10:
        print("Nota inválida. Debe ser entre 1 y 10.")
        nota = int(input("Ingrese nota (1-10): "))

    if nota >= 4:
        aprobados += 1
        estado = "Aprobó"
    else:
        desaprobados += 1
        estado = "Desaprobó"
        if nota == 1:
            aplazados += 1

    print(f"Alumno {legajo}: {estado} con nota {nota}")

total_alumnos = aprobados + desaprobados
porcentaje_aplazados = (aplazados * 100 / total_alumnos) if total_alumnos > 0 else 0

print(f"Cantidad de alumnos aprobados: {aprobados}")
print(f"Cantidad de alumnos desaprobados: {desaprobados}")
print(f"Porcentaje de alumnos aplazados (nota 1): {porcentaje_aplazados:.2f}%")
```
